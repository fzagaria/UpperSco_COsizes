```python
# Load Luhman22a's table:
luhman_stars = pd.read_fwf(
    "table1_luhman2022a_orig.dat",
    names  = [
        "SourceID_GaiaDR3", "SourceID_GaiaDR2", "NameOther",
        "RAdeg", "DEdeg",
        "SpType", "r_SpType", "AdoptSpType",
        "EWLi", "r_EWLi",
        "pmRA", "e_pmRA", "pmDec", "e_pmDec", "plx", "e_plx",
        "r_med_geo", "r_lo_geo", "r_hi_geo",
        "RVel", "e_RVel", "r_RVel",
        "U U", "e_U", "V V", "e_V", "W W", "e_W",
        "Gmag G", "e_Gmag", "GBPmag", "e_GBPmag", "GRPmag", "e_GRPmag",
        "RUWE", "kin", "pos",
        "2MASS", "SourceID_2MASS", "2MASS_angsep", "2MASS_closest",
        "WISE", "SourceID_WISE", "WISE_angsep", "WISE_closest"
    ],
    skiprows = 273,
)
print(len(luhman_stars))
# Numbers match: 10509, yes!
```

```python
# Load Ratzenbock+2023's table
ratzenbock = pd.read_csv(
    "tableE1_ratzenbock2023a.dat",
    header = None,
    names  = [
        "SourceID_GaiaDR3", "SigMA_membership", "Stability", "Distance (pc)", "e_d_upper (pc)", "e_d_lower (pc)",
        "X (pc)", "Y (pc)", "Z (pc)", "e_X_upper (pc)", "e_Y_upper (pc)", "e_Z_upper (pc)", "e_X_lower (pc)", "e_Y_lower (pc)", "e_Z_lower (pc)",
        "v_alpha (km s−1)", "v_delta (km s−1)", "e_v_alpha_upper (km s−1)", "e_v_delta_upper (km s−1)", "e_v_alpha_lower (km s−1)", "e_v_delta_lower (km s−1)",
        "v_alpha_LSR (km s−1)", "v_delta_LSR (km s−1)", "v_ERV (km s-1)",
    ],
    sep    = r"\s+",
)
print(len(ratzenbock))
# Numbers match (caption table E1): 13103.
```

```python
matched_clusters = luhman_stars.merge(
    ratzenbock[["SourceID_GaiaDR3","SigMA_membership"]],
    on  = ["SourceID_GaiaDR3"],
    how = "left"
)
print(len(matched_clusters))
matched_clusters["SourceID_GaiaDR3"] = matched_clusters["SourceID_GaiaDR3"].astype("string")
```

```python
tot = 0
for i in [1,2,3,4,5,6,7,8,9]:
    # Cluster - matching:
    print(f"Cluster {i}")
    matched_clusters_gaia = matched_clusters[matched_clusters["SigMA_membership"]==i]
    print(f"Number of members in cluster {i}:", len(matched_clusters_gaia))
    tot += len(matched_clusters_gaia)
print(f'Total Upper Sco stars = {tot}')

tot = 0
for i in [13]:
    # Cluster - matching:
    print(f"Cluster {i}")
    matched_clusters_gaia = matched_clusters[matched_clusters["SigMA_membership"]==i]
    print(f"Number of members in cluster {i}:", len(matched_clusters_gaia))
    tot += len(matched_clusters_gaia)
print(f'Total Lupus stars = {tot}')
```

```python
# Load Luhman22b's table:
luhman_discs = pd.read_fwf(
    "table1_luhman2022b_orig.dat",
    names  = [
        "WISE", "SourceID_WISE_discs", "RAdeg","DEdeg", "2MASS", "SourceID_2MASS_discs", "2MASS_angsep",
        "Jmag 2MASS", "e_Jmag", "Hmag 2MASS", "e_Hmag", "Ksmag 2MASS", "e_Ksmag",
        "W1mag", "e_W1mag", "W2mag", "e_W2mag", "W3mag", "e_W3mag", "f_W3mag", "W4mag", "e_W4mag", "f_W4mag",
        "ExcW2 Excess present in W2?", "ExcW3 Excess present in W3?", "ExcW4 Excess present in W4?", "DiskType Disk Typec"
    ],
    skiprows = 45,
)
print(len(luhman_discs))
# Numbers match: 9675, yes!
```

```python
# Remove those objects with no WISE counterpart in Luhman (2022a):
matched_clusters_match = matched_clusters.dropna(subset=["SourceID_WISE"])

# Match to add Gaia IDs, kin, pos info:
matched = luhman_discs.merge(
    matched_clusters_match[["SourceID_WISE", "WISE_closest", "SourceID_2MASS", "2MASS_closest", "SourceID_GaiaDR3", "SigMA_membership", "kin", "pos", "AdoptSpType"]],
    left_on  = ["SourceID_WISE_discs", "SourceID_2MASS_discs"],
    right_on = ["SourceID_WISE", "SourceID_2MASS"],
    how      = "left"
)
print(len(matched)) #9970

# For some weird reason there are some WISE IDs in Table 1 of Luhman (2022b) with no counterpart in Luhman (2022a)...
# These are matched to nan nan and I'll drop them since have no association with a Gaia source/kinpos info.
# WISEA  J121959.36-501839.6  184.9973407 -50.3110062 2MASS J12195937-5018380
# is present in Luhman (2022a) as
# WISEA  J121959.36-501839.6  184.9973407 -50.3110062 2MASS J12195938-5018404
# and I updated it accordingly in Luhman (2022b)
num_dropped = matched[["SourceID_WISE","SourceID_2MASS"]].isna().all(axis=1).sum()
print("WISE IDs in Table 1 of Luhman (2022b) with no counterpart in Table 1 of Luhman (2022a):", num_dropped)

matched = matched.dropna(subset=["SourceID_WISE","SourceID_2MASS"],how="all")
print(len(matched)) #9955

# Duplicates?
dups = matched[matched["SourceID_WISE"].duplicated(keep=False)]
print(len(np.unique(dups["SourceID_WISE"].tolist())))
# There are 295 duplicated sources (i.e., more Gaia sources are associated to the same WISE ref ID).
# Those share the same WISE ID - 2MASS ID combination is present in Table1 of Luhman (2022b)

# Perfect 'cause 9970 - 295 is 9675, i.e., the size of Table 1 of Luhman (2022b).
# with open("matched.dat", "w") as f:
#     f.write(matched.to_string(index=False))
```

```python
# The membership of 13 Upper Sco targets was re-evaluated by Carpenter et al. (2025) based on their ALMA data.
# The following are likely to be background objects and will be removed:
excludeobjs2MASS = ["J15442550-2126408", "J15564244-2039339", "J15594231-2945495", "J16044876-1748393", "J16050647-1734020", "J16153220-2010236", "J16185382-2053182", "J15524851-2845369", "J15594460-2155250", "J16052556-2035397", "J16064385-1908056", "J16120505-2043404", "J16185277-2259537"]

mask    = matched["SourceID_2MASS"].isin(excludeobjs2MASS)
# matched[mask]["SourceID_2MASS"]
matched = matched[~mask].copy()
print(len(matched["SourceID_2MASS"]))
```

```python
tot_both, tot_sing, tot_discs_both, tot_discs_sing = 0, 0, 0, 0
for i in [1,2,3,4,5,6,7,8,9]:
    # Cluster - matching:
    print(f"************************************")
    print(f"Cluster {i}")
    matched_clusters_wise = matched[matched["SigMA_membership"]==i]
    print(f"Number of members in cluster {i}:", len(matched_clusters_wise))
    
    # Single targets:
    sing_cluster = matched_clusters_wise[~matched_clusters_wise["SourceID_WISE_discs"].duplicated(keep=False)]
    # Duplicate targets:
    dups_cluster = matched_clusters_wise[ matched_clusters_wise["SourceID_WISE_discs"].duplicated(keep=False)]
    print(f"Check that length single+duplicates matches table: {len(sing_cluster)+len(dups_cluster)}")
    print(f"Length duplicates: {len(np.unique(dups_cluster["SourceID_WISE_discs"]))}")
    
    # Duplicate disc less: keep both stars (none has a disc)
    dups_cluster_discless    = dups_cluster[ dups_cluster["DiskType Disk Typec"].isin(excluded_class)]
    print(f"Length duplicates disc less   : {len(np.unique(dups_cluster_discless["SourceID_WISE_discs"]))}")
    # Duplicate disc bearing: consider both extreme scenarios - both have discs or closest match only)
    dups_cluster_discbearing = dups_cluster[~dups_cluster["DiskType Disk Typec"].isin(excluded_class)]
    print(f"Length duplicates disc bearing: {len(np.unique(dups_cluster_discbearing["SourceID_WISE_discs"]))}")
    print(
        f"Length duplicates closest disc bearing: {len(dups_cluster_discbearing[dups_cluster_discbearing["WISE_closest"]=="Y"])}"
    )
    
    # Make a table with the assumption that all the disc-bearing duplicates have both components disc-bearing
    cluster_both = pd.concat([sing_cluster, dups_cluster_discless, dups_cluster_discbearing], ignore_index=True)
    print(f"Number of stars both  : {len(cluster_both)}")
    # Make a table with the assumption that all the disc-bearing duplicates have one component disc-bearing (the closest)
    cluster_sing = pd.concat([sing_cluster, dups_cluster_discless, dups_cluster_discbearing[dups_cluster_discbearing["WISE_closest"]=="Y"]], ignore_index=True)
    print(f"Number of stars single: {len(cluster_sing)}")
    
    # Select disc-bearing stars in each catalogue
    print(f"Number of stars w/ discs both  : {len(cluster_both[~cluster_both["DiskType Disk Typec"].isin(excluded_class)])}")
    print(f"Number of stars w/ discs single: {len(cluster_sing[~cluster_sing["DiskType Disk Typec"].isin(excluded_class)])}")

    tot_both += len(cluster_both)
    tot_sing += len(cluster_sing)
    tot_discs_both += len(cluster_both[~cluster_both["DiskType Disk Typec"].isin(excluded_class)])
    tot_discs_sing += len(cluster_sing[~cluster_sing["DiskType Disk Typec"].isin(excluded_class)])
print(f"************************************")
print(f"Total number of stars in WISE in Upper Sco (both):", tot_both)
print(f"Total number of stars in WISE in Upper Sco (sing):", tot_sing)
print(f"Total number of stars w/ discs in WISE in Upper Sco (both):", tot_discs_both)
print(f"Total number of stars w/ discs in WISE in Upper Sco (sing):", tot_discs_sing)
```
